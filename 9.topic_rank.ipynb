```python
import pandas as pd
```

```python
df_local = pd.read_excel('./analysis-data/local_topic-freq.xlsx')
df_local['source'] = 'Local'
df_non = pd.read_excel('./analysis-data/non_topic-freq.xlsx')
df_non['source'] = 'Non-local'
```

```python
local_rank = df_local.copy()
non_rank = df_non.copy()
```

```python
local_rank = local_rank.sort_values('freq', ascending=False).reset_index(drop=True)
local_rank['rank_local'] = local_rank.index + 1
```

```python
local_rank
```

```python
non_rank = non_rank.sort_values('freq', ascending=False).reset_index(drop=True)
non_rank['rank_non'] = non_rank.index + 1
```

```python
non_rank
```

```python
local_rank = local_rank[["topic_en", "topic", "freq", "rank_local"]]
non_rank = non_rank[["topic_en", "topic", "freq", "rank_non"]]
df_rank_compare = pd.merge(
    local_rank,
    non_rank,
    on='topic_en',
    how='outer',
    suffixes=('_local', '_non')
)
```

```python
df_rank_compare
```

```python
df_rank_compare['rank_diff'] = df_rank_compare['rank_local'] - df_rank_compare['rank_non']
```

```python
df_rank_compare['abs_rank_diff'] = df_rank_compare['rank_diff'].abs()
```

```python
df_rank_compare = df_rank_compare.sort_values('abs_rank_diff', ascending=False)
df_rank_compare
```

```python
def output_rank(localdf, nondf):
    local = localdf.copy()
    non = nondf.copy()
   
    local = local.sort_values('freq', ascending=False).reset_index(drop=True)
    local['RANK'] = local.index + 1

    non = non.sort_values('freq', ascending=False).reset_index(drop=True)
    non['RANK'] = non.index + 1

    temp = pd.concat([local, non])
    temp[['topic_en', 'source', 'RANK']].to_excel('./analysis-data/topic_rank_diff.xlsx', index=None)
    return temp
```

```python
output_rank(df_local, df_non)
```
