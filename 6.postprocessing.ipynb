```python
import pandas as pd
```

```python
def clean_label(x):        
    labels = str(x).split('@')
    labels = [i.strip() for i in labels if i.strip()]  
    if '其他' in labels and len(labels) > 1:
        print(labels)
        labels = [i for i in labels if i != '其他']
    
    
    return '@'.join(labels)
```

```python
local =  pd.read_csv('./data/topic-results/local_topic.csv')
local['label'] = local['label'].fillna('其他')
local['label'] = local['label'].apply(clean_label)
local.to_csv('./data/topic-results/result_local_topic.csv', index=None)
```

```python
non_df = pd.read_csv('./data/topic-results/non_local.csv')
non_df['label'] = non_df['label'].fillna('其他')
non_df['label'] = non_df['label'].apply(clean_label)
non_df.to_csv('./data/topic-results/result_non_local.csv', index=None)
```
