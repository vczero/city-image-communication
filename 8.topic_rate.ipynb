```python
import pandas as pd

LEN_LOCAL = len(pd.read_csv('./data/topic-results/result_local_topic.csv'))
LEN_NON = len(pd.read_csv('./data/topic-results/result_non_local.csv'))

df_local = pd.read_excel('./analysis-data/local_topic-freq.xlsx')
df_local['source'] = 'Local'
df_non = pd.read_excel('./analysis-data/non_topic-freq.xlsx')
df_non['source'] = 'Non-local'
```

```python
def get_rate(df, _type):
    if _type == 'LOCAL':
        total = LEN_LOCAL
    else:
        total = LEN_NON

    print(total)
    df['percentage'] = df['freq'] / total * 100
    df['percentage'] = df['percentage'].round(2)
    return df
```

```python
temp = pd.concat([get_rate(df_local, 'LOCAL'), get_rate(df_non, 'NON')])
temp.to_excel('./analysis-data/lines_TopicDistribution.xlsx', index=None)
temp.to_csv('./analysis-data/lines_TopicDistribution.csv', index=None)
```
